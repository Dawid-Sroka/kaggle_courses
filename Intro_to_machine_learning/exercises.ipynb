# Course step 2 - Basic Data Exploration

```python
import pandas as pd

# load and review the data
iowa_file_path = "train.csv"
raw_data = pd.read_csv(iowa_file_path)
        
raw_data.describe()
```

# Course step 3 - Your First Machine Learning Model

```python
columns = raw_data.columns
data = raw_data

y = data.SalePrice
feature_names = ["LotArea", "YearBuilt", "1stFlrSF", "2ndFlrSF", "FullBath", "BedroomAbvGr", "TotRmsAbvGrd"]

X = data[feature_names]
X.dropna(axis=0)
X.describe()
# X.head()
```

```python
# Create and fit model
from sklearn.tree import DecisionTreeRegressor

iowa_model = DecisionTreeRegressor(random_state=1)
iowa_model.fit(X, y)

# make predictions
print(iowa_model.predict(X))
print(iowa_model.predict(X.head()) == y.head())
```

# Course step 4 - Model Validation

```python
# split data in training and validation data. Use metrics to measure model accuracy
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

X_train, X_val, y_train, y_val = train_test_split(X, y, random_state=1)
y_train_list = list(y_train)

iowa_model.fit(X_train, y_train)

predictions_in_sample = iowa_model.predict(X_train)
print(mean_absolute_error(y_train, predictions_in_sample))
print(predictions_in_sample[:5])
print(y_train_list[:5])
# for the top few values, the model is perfectly fitted
# however we see the mea is not zero even, so there are some predictions that are not perfect:
print([(i, predictions_in_sample[i], y_train_list[i]) for i in range(len(y_train)) if predictions_in_sample[i] != y_train_list[i]])

predictions_out_sample = iowa_model.predict(X_val)
print(predictions_out_sample[:5])
print(list(y_val)[:5])
print(mean_absolute_error(y_val, predictions_out_sample))
# we see mea is unacceptably high
```

# Course step 5: Underfitting and Overfitting

```python
n = len(X)
d = iowa_model.get_depth()
L = iowa_model.get_n_leaves()
print(n)
print(d)
print(L)
from math import log
print(log(1500))
```

```python
# let's try different numbers of leafs (the default is )
for i in [5, 25, 50, 100, 200, 500, 1000]:
    print()
    print("max_leaf_nodes = ", i)
    parametrized_iowa_model = DecisionTreeRegressor(max_leaf_nodes=i, random_state=2)
    parametrized_iowa_model.fit(X_train, y_train)

    predictions_in_sample = parametrized_iowa_model.predict(X_train)
    print("mea in  sample = ", mean_absolute_error(y_train, predictions_in_sample))
    # print(predictions_in_sample[:5])
    # print(y_train_list[:5])

    # for the top few values, the model is perfectly fitted
    # however we see the mea is not zero even, so there are some predictions that are not perfect:
    # print([(i, predictions_in_sample[i], y_train_list[i]) for i in range(len(y_train)) if predictions_in_sample[i] != y_train_list[i]])

    predictions_out_sample = parametrized_iowa_model.predict(X_val)
    print("mea out sample = ", mean_absolute_error(y_val, predictions_out_sample))
    # print(predictions_out_sample[:5])
    # print(list(y_val)[:5])
```

# Course Step 6: Random forest

```python
from sklearn.ensemble import RandomForestRegressor

forest_model = RandomForestRegressor(random_state=1)
forest_model.fit(X_train, y_train)

predictions_out_sample = forest_model.predict(X_val)
mae = mean_absolute_error(y_val, predictions_out_sample)
print(mae)

# I got 21857, which is better then 27467 I got from DecisionTree with optimal number of leafs
```
